```python
import numpy as np
```

```python
X = (0, 2)
Y = (0, 1)

M = lambda x, y: 3*x + y
```

```python
x_vals = np.linspace(X[0], X[1], 100)
y_vals = np.linspace(Y[0], Y[1], 100)

min_for_x = [min(M(x, y) for y in y_vals) for x in x_vals]
maxmin = max(min_for_x)

max_for_y = [max(M(x, y) for x in x_vals) for y in y_vals]
minmax = min(max_for_y)

x_maxmin, y_maxmin = x_vals[np.argmax(min_for_x)], y_vals[np.argmax(min_for_x)]
x_minmax, y_minmax = x_vals[np.argmin(max_for_y)], y_vals[np.argmin(max_for_y)]

print(f'Maxmin = {maxmin}, x = {x_maxmin}, y = {y_maxmin}')
print(f'Minmax = {minmax}, x = {x_minmax}, y = {y_minmax}')
```

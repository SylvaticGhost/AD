```python
import numpy as np
```

```python
M = [
    [12, 4, 2, 6, 12],
    [12, 4, 10, 2, 16],
    [4, 13, 14, 8, 5],
    [12, 11, 3, 9, 15],
    [1, 3, 4, 2, 10]
]

a = 0.25
```

```python
H = []

for strategy in M:
    H_i = a * np.min(strategy) + (1 - a) * np.max(strategy)
    H.append(H_i)

print(f'H_i для стратегій: {H}')
```

```python
max_H = np.max(H)
i = H.index(max_H)
print(f'Найкраща стратегія: {i + 1}')
```

## Оцінка параметрів

```python
df = pd.read_csv('dkr6_data.csv')
df
```

* Варітсть: 0 -дорого, 1 -недорого
* Розмір: 0 -малий, 1 - середній, 2 - великий
* Стан: 0 - капітальний ремонт, 1 - потребує косметичний ремонт, 2 - проведено капітальний ремонт
* Відстань: 0 - поза містом, 1 - далеко від центру, 2 - недалеко

```python
criteria = ["Вартість", "Розмір", "Стан", "Відстань"]
data = df[criteria].values

def dominates(a, b):
    return all(a_i >= b_i for a_i, b_i in zip(a, b)) and any(a_i > b_i for a_i, b_i in zip(a, b))

pareto = []
for i, row_i in enumerate(data):
    if not any(dominates(other, row_i) for j, other in enumerate(data) if j != i):
        pareto.append(i)

pareto_df = df.iloc[pareto]
print("Множина Еджворта-Парето:")
pareto_df
```

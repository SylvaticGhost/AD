```python
import pandas as pd
import numpy as np
```

```python
def to_proccents(x):
    return f"{x * 100:.4f}%"
```

```python
A = {
    0.15: 0.25,
    0.17: 0.5,
    0.1: 0.25
}

B = {
    0.14: 0.23,
    0.12: 0.43,
    0.10: 0.34
}

C = {
    0.05: 0.75,
    0.13: 0.15,
    0.2: 0.1
}
```

```python
def create_df(data):
    return pd.DataFrame(list(data.items()), columns=['Return', 'Probability']).set_index('Return').sort_index(ascending=False)

df_A = create_df(A)
df_B = create_df(B)
df_C = create_df(C)
```

```python
def analyze_company(df, name):
    expected_return = (df.index * df['Probability']).sum()
    variance = ((df.index - expected_return) ** 2 * df['Probability']).sum()
    std = np.sqrt(variance)
    cv = std / expected_return if expected_return != 0 else np.nan
    print(f"Компанія {name}:")
    print(f"Очіквана прибутковість: {to_proccents(expected_return)}")
    print("Дисперсія:", variance)
    print("Стандартне відхилення:", to_proccents(std))
    print("Коефіцієнт варіації:", cv)
```

```python
analyze_company(df_A, 'A')
```

```python
analyze_company(df_B, 'B')
```

```python
analyze_company(df_C, 'C')
```

* Компанія A - гарний варіант для інвестора з помірним ризиком
* Компанія B - найкраще співвідношення ризику та прибутковості(найнижчий коефіцієнт варіації), чудовий варіант для тих хто не хоче сильно ризикувати, але в тойже час отримати не поганий дохід
* Компанія C - найвищий ризик, низький дохід найгігрший ризик
